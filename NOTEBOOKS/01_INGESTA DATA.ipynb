```python
# --- 6. Carga de Datos (CSV) : PROCESSED_GL_DATA_ALL_P_L_ANNUAL_PER_ENTERPRISE ---
import pandas as pd
import os
import logging

# Configuración básica de logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

# --- 5. Carga de Datos (CSV) y Generación de P&L POR AÑO Y EMPRESA ---
logging.info("Ingesta & Preprocesamiento: Iniciando carga y cálculo de P&L por Empresa/Portafolio.")

# 1. Definir la ruta del archivo de entrada
file_path = '../DATA/RAW/GL_DATA_ALL.CSV'

try:
    # Usa pd.read_csv para leer el archivo CSV. Es crucial usar header=None
    # ya que se asume que las columnas no tienen nombre inicial
    df = pd.read_csv(file_path, header=None)
    logging.info(f"Datos cargados exitosamente desde '{file_path}'. Dimensiones: {df.shape}")

    # --- RENOMBRAR COLUMNAS ---
    logging.info("GL_DATA_ALL.CSV: Asigna Título a Columnas")
    mapeo_columnas = {
        0: 'LIBDIARIOID', 1: 'ENTERPRISEID', 2: 'ENTEPRISENAME', 3: 'PROPERTYID',
        4: 'LIBDIATYPEID', 5: 'YEARID', 6: 'MONTHID', 7: 'SOURCEID',
        8: 'ODATAID', 9: 'COACODDETAIL', 10: 'DESCTYPE', 11: 'DESCSUBTYPE',
        12: 'COANAME', 13: 'DEBIT', 14: 'CREDIT', 15: 'NET',
        16: 'LOCATION', 17: 'STATE', 18: 'CITY', 19: 'STATEID',
        20: 'DESC_TYPENN', 21: 'DESC_TYPEPROPERTY', 22: 'MJ_OWNER', 23: 'OPT_LEASE',
        24: 'CAPRATE', 25: 'TENANTID', 26: 'TENANTNAME', 27: 'DESC_TYPEMGMT',
        28: 'PORTFOLIOID', 29: 'PORTFOLIONAME', 30: 'ADMIN_ENTITYID', 31: 'ADMIN_ENTITY',
        32: 'DESC_TYPETENANT', 33: 'DESC_CATEGORY'
    }
    # Aplicar el renombramiento
    df_renombrado = df.rename(columns=mapeo_columnas)

    # --- CALCULO DE NET INCOME (P&L POR AÑO, EMPRESA Y PORTAFOLIO) ---
    logging.info("Aplicando filtros y agrupando para el cálculo de Net Income.")

    # CRITERIOS DE FILTRADO (Se mantienen los criterios del P&L global)
    # 1. YEARID entre 2020 y 2024
    filtro_anio = (df_renombrado['YEARID'] >= 2020) & (df_renombrado['YEARID'] <= 2024)
    
    # 2. DESCTYPE igual a 'Profit & Loss'
    filtro_pl = df_renombrado['DESCTYPE'] == 'Profit & Loss'
    
    # 3. LIBDIATYPEID del valor 2 al 18 (inclusive)
    filtro_libdia = (df_renombrado['LIBDIATYPEID'] >= 2) & (df_renombrado['LIBDIATYPEID'] <= 18)

    # Aplicar todos los filtros
    df_pl_filtrado = df_renombrado[filtro_anio & filtro_pl & filtro_libdia].copy()
    
    # Verificar si el DataFrame filtrado no está vacío
    if df_pl_filtrado.empty:
        logging.warning("El DataFrame filtrado para P&L está vacío. No se puede calcular el Net Income.")
        exit()

    # Agrupar por las nuevas dimensiones y sumar el 'NET'
    grupos_por_empresa_portfolio = [
        'YEARID', 
        'ENTERPRISEID', 
        'ENTEPRISENAME', 
        'PORTFOLIOID', 
        'PORTFOLIONAME'
    ]
    
    df_net_income = df_pl_filtrado.groupby(grupos_por_empresa_portfolio)['NET'].sum().reset_index()
    
    # Renombrar la columna del resultado y redondear
    df_net_income.rename(columns={'NET': 'NETINCOME'}, inplace=True)
    df_net_income['NETINCOME'] = df_net_income['NETINCOME'].round(2)

    # Ordenar por año y empresa para mejor visualización
    df_net_income = df_net_income.sort_values(by=['YEARID', 'ENTERPRISEID'])

    print("\n--- PRIMERAS FILAS DEL NET INCOME (P&L) POR EMPRESA/PORTAFOLIO ---")
    print(df_net_income.head())
    print("---------------------------------------------------------------------")

    # --- 6. Guardar el Resultado Final ---
    # Se ajusta el nombre del archivo para reflejar la granularidad (Per Enterprise)
    nombre_archivo = 'PROCESSED_GL_DATA_ALL_P_L_ANNUAL_PER_ENTERPRISE.csv'
    carpeta_destino = '../DATA/PROCESSED'
    ruta_completa = os.path.join(carpeta_destino, nombre_archivo)

    # Crear la carpeta de destino si no existe
    os.makedirs(carpeta_destino, exist_ok=True)
    
    # Guardar el DataFrame con el P&L anual por empresa/portafolio
    df_net_income.to_csv(
        ruta_completa,
        index=False,
        encoding='utf-8',
        sep=','
    )
    
    print(f"\n✅ ¡Guardado exitoso! El P&L por Empresa/Portafolio se encuentra en: {ruta_completa}")
    logging.info("Generación de dataset de P&L: PROCESSED_GL_DATA_ALL_P_L_ANNUAL_PER_ENTERPRISE.csv")

except FileNotFoundError:
    print(f"\n❌ Error: El archivo '{file_path}' no fue encontrado.")
    print("Asegúrate de que el archivo 'GL_DATA_ALL.CSV' esté en la carpeta '../DATA/RAW/'.")
except pd.errors.EmptyDataError:
    print(f"\n❌ Error: El archivo '{file_path}' está vacío.")
except Exception as e:
    print(f"\n❌ Ocurrió un error inesperado al procesar el archivo: {e}")
```

```python
# ---10. Carga de Datos (CSV) : PROCESSED_GL_DATA_ALL_P_L_ANNUAL_MONTHLY_PER_ENTERPRISEV5 ---
import pandas as pd
import os
import logging

# Configuración básica de logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

# --- 5. Carga de Datos (CSV) y Generación de P&L POR AÑO Y EMPRESA ---
logging.info("Ingesta & Preprocesamiento: Iniciando carga y cálculo de P&L por Empresa/Portafolio.")

# 1. Definir la ruta del archivo de entrada
file_path = '../DATA/RAW/GL_DATA_ALL.CSV'

try:
    # Usa pd.read_csv para leer el archivo CSV. Es crucial usar header=None
    # ya que se asume que las columnas no tienen nombre inicial
    df = pd.read_csv(file_path, header=None)
    logging.info(f"Datos cargados exitosamente desde '{file_path}'. Dimensiones: {df.shape}")

    # --- RENOMBRAR COLUMNAS ---
    logging.info("GL_DATA_ALL.CSV: Asigna Título a Columnas")
    mapeo_columnas = {
        0: 'LIBDIARIOID', 1: 'ENTERPRISEID', 2: 'ENTEPRISENAME', 3: 'PROPERTYID',
        4: 'LIBDIATYPEID', 5: 'YEARID', 6: 'MONTHID', 7: 'SOURCEID',
        8: 'ODATAID', 9: 'COACODDETAIL', 10: 'DESCTYPE', 11: 'DESCSUBTYPE',
        12: 'COANAME', 13: 'DEBIT', 14: 'CREDIT', 15: 'NET',
        16: 'LOCATION', 17: 'STATE', 18: 'CITY', 19: 'STATEID',
        20: 'DESC_TYPENN', 21: 'DESC_TYPEPROPERTY', 22: 'MJ_OWNER', 23: 'OPT_LEASE',
        24: 'CAPRATE', 25: 'TENANTID', 26: 'TENANTNAME', 27: 'DESC_TYPEMGMT',
        28: 'PORTFOLIOID', 29: 'PORTFOLIONAME', 30: 'ADMIN_ENTITYID', 31: 'ADMIN_ENTITY',
        32: 'DESC_TYPETENANT', 33: 'DESC_CATEGORY'
    }
    # Aplicar el renombramiento
    df_renombrado = df.rename(columns=mapeo_columnas)

    # --- CALCULO DE NET INCOME (P&L POR AÑO, EMPRESA Y PORTAFOLIO) ---
    logging.info("Aplicando filtros y agrupando para el cálculo de Net Income.")

    # CRITERIOS DE FILTRADO (Se mantienen los criterios del P&L global)
    # 1. YEARID entre 2020 y 2024
    filtro_anio = (df_renombrado['YEARID'] >= 2020) & (df_renombrado['YEARID'] <= 2024)
    
    # 2. DESCTYPE igual a 'Profit & Loss'
    filtro_pl = df_renombrado['DESCTYPE'] == 'Profit & Loss'
    
    # 3. LIBDIATYPEID del valor 2 al 18 (inclusive)
    filtro_libdia = (df_renombrado['LIBDIATYPEID'] >= 2) & (df_renombrado['LIBDIATYPEID'] <= 18)

    # Aplicar todos los filtros
    df_pl_filtrado = df_renombrado[filtro_anio & filtro_pl & filtro_libdia].copy()
    
    # Verificar si el DataFrame filtrado no está vacío
    if df_pl_filtrado.empty:
        logging.warning("El DataFrame filtrado para P&L está vacío. No se puede calcular el Net Income.")
        exit()

    # Agrupar por las nuevas dimensiones y sumar el 'NET'
    grupos_por_empresa_portfolio = [
        'YEARID', 
        'MONTHID',
        'PORTFOLIOID', 
        'PORTFOLIONAME'
    ]
    
    df_net_income = df_pl_filtrado.groupby(grupos_por_empresa_portfolio)['NET'].sum().reset_index()
    
    # Renombrar la columna del resultado y redondear
    df_net_income.rename(columns={'NET': 'NETINCOME'}, inplace=True)
    df_net_income['NETINCOME'] = df_net_income['NETINCOME'].round(2)

    # ❗ APLICACIÓN DE LA CORRECCIÓN SOLICITADA: Multiplicar el resultado del Net Income por -1
    logging.info("Aplicando corrección: NETINCOME = NETINCOME * -1.")
    df_net_income['NETINCOME'] = df_net_income['NETINCOME'] * -1
    # -----------------------------------------------------------------------------------------

    # --- NUEVA LÓGICA: CREAR PKRECORD (YEARID + MONTHID + ENTERPRISEID) ---
    logging.info("Generando la clave compuesta 'PKRECORD'.")
    # Convertir a string explícitamente y luego concatenar
    df_net_income['PKRECORD'] = (
        df_net_income['YEARID'].astype(str) + 
        '_' + 
        df_net_income['MONTHID'].astype(str).str.zfill(2) +  # Asegura dos dígitos para el mes (ej: 01, 05)
        '_' + 
        df_net_income['PORTFOLIOID'].astype(str)
    )
    # Reordenar las columnas para que PKRECORD sea la primera (opcional, pero buena práctica)
    columnas = ['PKRECORD'] + [col for col in df_net_income.columns if col != 'PKRECORD']
    df_net_income = df_net_income[columnas]
    # --------------------------------------------------------------------

    # Ordenar por año y empresa para mejor visualización
    df_net_income = df_net_income.sort_values(by=['YEARID', 'MONTHID'])

    print("\n--- PRIMERAS FILAS DEL NET INCOME (P&L) POR EMPRESA/PORTAFOLIO ---")
    print(df_net_income.head())
    print("---------------------------------------------------------------------")

    # --- 6. Guardar el Resultado Final ---
    # Se ajusta el nombre del archivo para reflejar la granularidad (Per Enterprise)
    nombre_archivo = 'PROCESSED_GL_DATA_ALL_P_L_ANNUAL_MONTHLY_PER_ENTERPRISEV6.csv'
    carpeta_destino = '../DATA/PROCESSED'
    ruta_completa = os.path.join(carpeta_destino, nombre_archivo)

    # Crear la carpeta de destino si no existe
    os.makedirs(carpeta_destino, exist_ok=True)
    
    # Guardar el DataFrame con el P&L anual por empresa/portafolio
    df_net_income.to_csv(
        ruta_completa,
        index=False,
        encoding='utf-8',
        sep=','
    )
    
    print(f"\n✅ ¡Guardado exitoso! El P&L por Empresa/Portafolio se encuentra en: {ruta_completa}")
    logging.info("Generación de dataset de P&L: PROCESSED_GL_DATA_ALL_P_L_ANNUAL_MONTHLY_PER_ENTERPRISEV6.csv")

except FileNotFoundError:
    print(f"\n❌ Error: El archivo '{file_path}' no fue encontrado.")
    print("Asegúrate de que el archivo 'GL_DATA_ALL.CSV' esté en la carpeta '../DATA/RAW/'.")
except pd.errors.EmptyDataError:
    print(f"\n❌ Error: El archivo '{file_path}' está vacío.")
except Exception as e:
    print(f"\n❌ Ocurrió un error inesperado al procesar el archivo: {e}")
```
