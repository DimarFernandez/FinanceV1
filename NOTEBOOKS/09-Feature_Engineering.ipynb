### Feature Engineering

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')


je_data = pd.read_csv('PROCESSED_GL_DATA_ALL.csv')
 
```

```python
je_data.sample(3).T
```

```python
#Fill categorical
print("La forma del préstamo_demográfico es {}".format(je_data.shape))
je_data.isna().sum()
```

```python
je_data['PORTFOLIONAME'].value_counts()
```

```python
#fill with mode (Permanent)
je_data['PORTFOLIONAME'] = je_data['PORTFOLIONAME'].fillna(value='L2-ALL')
je_data.isna().sum()
```

```python
je_data['CAPRATE'].value_counts()
```

```python
#fill with mode (Permanent)
je_data['CAPRATE'] = je_data['CAPRATE'].fillna(value='7')
je_data.isna().sum()
```

```python
je_data['ADMIN_ENTITYID'].value_counts()
```

```python
#fill with mode (Permanent)
je_data['ADMIN_ENTITYID'] = je_data['ADMIN_ENTITYID'].fillna(value='6')
je_data.isna().sum()
```

```python
je_data['ENTERPRISEID'].value_counts()
```

```python
#fill with mode (Permanent)
je_data['ENTERPRISEID'] = je_data['ENTERPRISEID'].fillna(value='6')
je_data.isna().sum()
```

```python
je_data.isna().sum()
```

```python
mean_df = round(je_data['CAPRATE'].mean())
mode_df = round(je_data['CAPRATE'].mode()[0])
median_df = round(je_data['CAPRATE'].median())

#Fill with mean
print("CAPRATE with mean value of {}".format(mean_df))
je_data['CAPRATE'] = je_data['CAPRATE'].fillna(mean_df)

#Fill with mode
print("CAPRATE with modal value of {}".format(mode_df))
je_data['CAPRATE'] = je_data['CAPRATE'].fillna(mode_df)

#Fill with median
print("CAPRATE with median value of {}".format(median_df))
je_data['CAPRATE'] = je_data['CAPRATE'].fillna(median_df)
```

```python
#get correlated values
plt.figure(figsize = (15,10))
# Calculate correlation only for numeric columns
sns.heatmap(je_data.corr(numeric_only=True))
plt.show() # Remember to show the plot
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

to_train = ['PORTFOLIOID', 'CAPRATE','ADMIN_ENTITYID', 'ENTERPRISEID' ]

temp_df = je_data[to_train]
#Step-1: Split the dataset that contains the missing values and no missing values as test and train respectively.
x_train = temp_df[temp_df['PORTFOLIOID'].notnull()].drop(columns='PORTFOLIOID')
y_train = temp_df[temp_df['PORTFOLIOID'].notnull()]['PORTFOLIOID']
x_test = temp_df[temp_df['PORTFOLIOID'].isnull()].drop(columns='PORTFOLIOID')


lr.fit(x_train, y_train)
pred = lr.predict(x_test)

print(np.round(pred, 5))

je_data['PORTFOLIOID'][je_data['PORTFOLIOID'].isnull()] = pred
```

```python
#get the index of missing so we can some of the values used for filling
missing_indx = list(je_data['PORTFOLIOID'][je_data['PORTFOLIOID'].isna()].index)
```

```python
# explicitly require this experimental feature
from sklearn.experimental import enable_iterative_imputer  
from sklearn.impute import IterativeImputer

from sklearn.ensemble import RandomForestRegressor

# Run imputer with a Random Forest estimator
imp = IterativeImputer(RandomForestRegressor(n_estimators=5), max_iter=5, random_state=1)

to_train = ['PORTFOLIOID', 'CAPRATE','ADMIN_ENTITYID', 'ENTERPRISEID']
je_data[to_train] = pd.DataFrame(imp.fit_transform(je_data[to_train]), columns=to_train)
```

```python
je_data['PORTFOLIOID'][missing_indx].head(10)
```

```python
#The feature level_of_education is ordinal
je_data['PORTFOLIONAME'].unique()
```

```python
#use a simple map function
map_education = {"L2-ALL" : 1, "L2B": 2, "3RD": 3, "LJL": 4}
    
je_data['Portfolio_Entity'] = je_data['PORTFOLIONAME'].map(map_education)

je_data['Portfolio_Entity'].value_counts()
```

__Codificación Hash:__ La codificación hash, o hash de características, es una forma rápida y eficiente de codificar características. Es muy eficiente para características categóricas con un gran número de clases. El codificador hash funciona aplicando una función hash a las características.

```python
cat_cols = ['PORTFOLIONAME','ENTEPRISENAME'] 

hash_enc = ce.HashingEncoder(cols=cat_cols, n_components=10)
je_data = hash_enc.fit_transform(je_data)

je_data.head().T
```

```python
#Create feature interest
je_data['NETV2'] = je_data['DEBIT'] - je_data['CREDIT']
```

```python
# Groupby customer id and calculate their total loans taken
loannumber_count = (
    je_data.groupby(by='PORTFOLIOID')['ENTERPRISEID']
    .count() # Use .count() directly, which returns a Series
    .reset_index(name='LIBDIARIOID') # Convert Series back to DataFrame and name the new column
)

# merge back to dataset on customer_id
je_data = je_data.merge(right=loannumber_count, how='left', on='PORTFOLIOID')

# Verifica el resultado
je_data.head().T
```

```python
#create feature speed in sendy dataset
je_data['NETV3'] = je_data['DEBIT'] - je_data['CREDIT']
je_data.head().T
```

```python
#Use Sklearn Polynomial Features
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures()
to_cross = ['YEARID', 'MONTHID', 'ADMIN_ENTITYID']
crossed_feats = poly.fit_transform(je_data[to_cross].values)

#Convert to Pandas DataFrame and merge to original dataset
crossed_feats = pd.DataFrame(crossed_feats)
je_data = pd.concat([je_data, crossed_feats], axis=1)
```
